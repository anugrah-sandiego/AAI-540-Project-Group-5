# Exploratory Data Analysis

This notebook performs exploratory data analysis on the bank marketing dataset to understand the data distribution, relationships, and patterns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Add parent directory to path
import sys
sys.path.append(str(Path.cwd().parent))
```

```python
# Load data
data_path = '../data/raw/bank.csv'
df = pd.read_csv(data_path)

print(f"Dataset shape: {df.shape}")
print(f"\nDataset info:")
df.info()
```

```python
# Display first few rows
df.head()
```

```python
# Check for missing values
print("Missing values:")
df.isnull().sum()
```

```python
# Summary statistics
df.describe()
```

```python
# Target variable distribution
if 'y' in df.columns:
    plt.figure(figsize=(8, 5))
    sns.countplot(data=df, x='y')
    plt.title('Distribution of Target Variable (Term Deposit Subscription)')
    plt.xlabel('Subscribed')
    plt.ylabel('Count')
    plt.show()
    
    print(df['y'].value_counts(normalize=True))
```

```python
# Numerical features distribution
numeric_cols = df.select_dtypes(include=[np.number]).columns
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()

for idx, col in enumerate(numeric_cols[:6]):
    sns.histplot(data=df, x=col, kde=True, ax=axes[idx])
    axes[idx].set_title(f'Distribution of {col}')

plt.tight_layout()
plt.show()
```

```python
# Correlation matrix
plt.figure(figsize=(12, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numerical Features')
plt.show()
```

```python
# Categorical features analysis
categorical_cols = df.select_dtypes(include=['object']).columns
print(f"Categorical columns: {categorical_cols.tolist()}")
```

```python
# Categorical features vs target
if 'y' in df.columns:
    cat_cols_to_plot = categorical_cols[:4]
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    axes = axes.ravel()
    
    for idx, col in enumerate(cat_cols_to_plot):
        if col != 'y':
            sns.countplot(data=df, x=col, hue='y', ax=axes[idx])
            axes[idx].set_title(f'{col} vs Target')
            axes[idx].tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()
```

```python
# Outlier detection using boxplots
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.ravel()

for idx, col in enumerate(numeric_cols[:6]):
    sns.boxplot(data=df, y=col, ax=axes[idx])
    axes[idx].set_title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```
